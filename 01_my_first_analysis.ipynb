```python
# This loads the tools we need
import pandas as pd

# Test if it works
print("✅ Python is working!")
print("✅ pandas is ready!")
print("\n🎉 You're a data analyst now! Let's see some data...")
```

```python
df = pd.read_json('v2_token.json')

print(f"📊 We have {len(df)} currencies!")
print(f"📋 Dataset has {len(df.columns)} columns of information")

# Show the first 5 rows
df.head()
```

```python
print("=" * 60)
print("📝 INFORMATION AVAILABLE FOR EACH CURRENCY")
print("=" * 60)

for i, column in enumerate(df.columns, 1):
    print(f"{i:>2}. 🔹 {column}")

print("-" * 60)
print(f"📊 Total fields: {len(df.columns)} columns")
print("=" * 60)
```

```python
print("=" * 60)
print("🌐 WEBSITE AVAILABILITY ANALYSIS")
print("=" * 60)

has_website = df['website'].notna().sum()
total = len(df)
percentage = (has_website / total * 100) if total > 0 else 0

print(f"🔢 Total currencies        : {total}")
print(f"🌍 Currencies with website : {has_website}")
print(f"📊 Coverage                : {percentage:.1f}%")

print("-" * 60)
print("✨ EXAMPLES OF CURRENCIES WITH WEBSITES")
print("-" * 60)

examples = (
    df[df['website'].notna()][['symbol', 'base_currency_desc', 'website']]
    .head(5)
)

print(examples.to_string(index=False))

print("=" * 60)
```

```python
import matplotlib.pyplot as plt

top_currencies = df['symbol'].value_counts().head(10)

plt.figure(figsize=(10, 6))
plt.bar(top_currencies.index, top_currencies.values)
plt.xlabel('Cryptocurrency Symbol')
plt.ylabel('Count')
plt.title('Top 10 Currencies (JSON Dataset)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("🎉 Chart created successfully!")
```

```python
print("📊 QUICK STATS")
print("=" * 40)
print(f"Total currencies: {len(df)}")
print(f"Total columns: {len(df.columns)}")
print(f"With website: {df['website'].notna().sum()}")
print(f"With description: {df['base_currency_desc'].notna().sum()}")
print("=" * 40)

print("🚀 Task 1 completed using JSON!")
```
